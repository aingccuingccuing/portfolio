```python
class Dog:

    def __init__(self, name):
        self.name

    def bark(self):
        print('wof')
```

```python
def __init__(self,name):
    self.name = name

def bark(self,):
    print('bark')

# 그니까, 이렇게 포함할 기존 method들을 dict로 정의해놓으면
# type을 써서 새로운 객체를 만들 수 있다!!
d = {'__init__': __init__, 'bark':bark}

# 'Dog' the class, () no parent, d the dictionary of methods
Dog = type('Dog', (), d)
```

```python
#parent class

class Animal:
    def __init__(self, name):
        self.name = name

    def describe(self):
        # class 이름도 이런식으로 접근가능하다. cls느낌인가?
        # self의 class니까, self 라는 카테고리에 해당하는 class의 name이 출력된다
        # 하단을 보면, Animal이 아니라 Animal을 상속받는 개별 class들의 __name__ 메서드가 호출된다
        # __name__도 class의 이름,즉, type으로 동적 생성할 때 받는 바로 그 제일 처음의 str
        # type('Dog', ~~ 의 'Dog'가 바로 __name__이다!
        return f'this is {self.name}, a {self.__class__.__name__}'

def dog_init(self, name, breed):
    super(Dog, self).__init__(name)
    self.breed = breed

def cat_init(self, name, color):
    super(Cat, self).__init__(name)
    self.color = color

def bird_init(self, name, can_fly):
    super(Bird, self).__init__(name)
    self.can_fly = can_fly

# dynamic class 만들기
Dog = type('Dog', (Animal,), {'__init__': dog_init, 'bark': lambda self: print(f'{self.name} barks!')})

Cat = type('Cat', (Animal,), {'__init__': cat_init, 'meow': lambda self: print(f'{self.name} says Meow')})

Bird = type('Bird', (Animal,), {'__init__': bird_init, 'chirp': lambda self: print(f'{self.name} chirp chirp 호우')})

# 객체 생성 (개체들)
doggo = Dog('마음이', '진돗개')
냥냥이 = Cat('나비', '삼색')
엥 = Bird('고라니', False)

# methods of each instance
print(doggo.describe())
doggo.bark()

print(냥냥이.describe())
냥냥이.meow()

print(엥.describe())
엥.chirp()
```

# 9) `__slots__`

```python
class Dog:
    # slots로 class가 받을 수 있는 변수들, attributes를 제한할 수 있다!!
    __slots__ = ['name', 'age']

dog = Dog()

dog.name = '아담'
dog.age = 932

dog.gender = 'male'
```

```python

# 심지어 다른 내장함수들도 제한되어 버린다
print(dog.__dict__)
```

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def noise(self):
        pass
```

```python
class Dog(Animal):
    def __init__(self):
        pass

    def noise(self):
        print("멍멍")
```

```python
dog = Dog()
dog.noise()
```
